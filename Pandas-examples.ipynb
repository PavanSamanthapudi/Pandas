```python
# Create Series from NumPy array
import numpy as np
arr = np.array([5, 10, 15])
s4 = pd.Series(arr, index=['x', 'y', 'z'])
print(s4)
```

```python
#  Series Attributes & Properties

s2 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'], name='Scores')

print(s2)
print(s2.index)     # Index object
print(s2.values)    # NumPy array of values
print(s2.dtype)     # Data type
print(s2.name)      # Name of the Series
print(s2.shape) 
type(s2.shape)
```

#### Series Indexing & Access

```python
# 1. Positional Indexing
# [10 20 30 40]
print(s2.values[0])
print(s2.values[3])
```

```python
# iloc gets rows (and/or columns) at integer locations.

print(s2.iloc[0])
```

```python
print(s2.iloc[1:3])
```

```python
# 2. Label-based Indexing

# loc gets rows (and/or columns) with particular labels.
# Pandas resolves s2[0] in the following order:

# Check if 0 is a label in the index
# → If yes, return by label

# If not, and if the index is not integer-based, then treat 0 as a position
# → Return item at index position 0 (first element)

# FutureWarning: Series.__getitem__ treating keys as positions is deprecated.

# In a future version, integer keys will always be treated as 
# labels (consistent with DataFrame behavior). 
# To access a value by position, use `ser.iloc[pos]`

print(s2.values[0])
```

```python
print(s2.iloc[0])
```

```python
print(s2.iloc[1:3])
```

```python
print(s2.loc['b'])
```

```python
print(s2.loc[['b', 'd']])
```

```python
s3 = pd.Series([10,20,30,40,50,60])
print(s3.loc[1])
print(s3.loc[[1,2]]) # loc will work for any label , be it numberic or alpha index. But iloc have to be used only for numeric index
```
